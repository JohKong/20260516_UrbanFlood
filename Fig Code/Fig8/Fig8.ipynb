```python
# 例如：你想选第 1, 10, 50, 100 这四个时刻


df = pd.read_csv('Run12/plotLine1.csv')

print(df.head())

df.rename(columns={'Points:0':'x'}, inplace=True)

print(df.head())

selected_times = [1, 10, 50, 100]
filtered_df = df[df['Time'].isin(selected_times)]

plt.rcParams.update({
    'font.size': 14,
    'font.family': 'Times New Roman',
    "text.usetex": True
})

# 4. 绘图 - 使用子图对象
fig, ax = plt.subplots(1, figsize=(6, 5))

# 颜色循环，使用更专业的调色板
colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728']

# 遍历每个选定的时刻进行画线
for i, t in enumerate(selected_times):
    subset = filtered_df[filtered_df['Time'] == t]
    y_col = df.columns[1] 
    
    # 绘制点线图
    ax.plot(
        subset['x'], 
        subset[y_col], 
        linewidth=1.2,
        linestyle='-',
        marker='x',
        markersize=6,
        markerfacecolor=colors[i],
        markeredgecolor=colors[i],
        markeredgewidth=0.8,
        color=colors[i],
        markevery=20,  # 每5个点显示一个标记，避免过于密集
        label=rf'$t={t}s$'
    )

# 添加灰色区域
ax.axvspan(0.65, 1.45, alpha=0.1, facecolor='#7f7f7f', edgecolor='black', lw=1.5, ls='-')

# 5. 图表修饰 - 按照示例样式调整
ax.set_xlabel(r'$x\;(m)$')
ax.set_ylabel(r'$TKE\;(m^2/s^2)$')
#ax.set_title(r'Turbulence intensity of LCZ10 at $y=0.5m$ and $z=0.0242m$')

ax.set_ylim(0, 0.11)

# 设置坐标轴样式
ax.xaxis.set_tick_params(direction='in', which='both')
ax.yaxis.set_tick_params(direction='in', which='both')

# 设置网格
ax.grid(True, linestyle='--', linewidth=0.5, alpha=0.7)

ax.text(0.26, 0.5, 'LCZ10 Area', transform=ax.transAxes, 
         fontsize=14, fontweight='bold', va='top')

ax.text(
            -0.12, 1.05, '(d)',
            transform=ax.transAxes,
            fontsize=18,
            fontweight='bold',
            va='top',
            ha='left'
        )


# 设置边框样式
for spine in ax.spines.values():
    spine.set_linewidth(1.5)
    spine.set_color('black')

# 调整图例位置和样式
ax.legend(
    frameon=True,
    loc='best',
    fontsize=14
)

# 如果需要坐标轴在顶部和左侧
# ax.yaxis.tick_left()
# ax.xaxis.tick_top()

plt.tight_layout()
plt.savefig('TKE(d).png', dpi=600, bbox_inches='tight')
plt.show()
```

```python
from PIL import Image

def merge_2x2_exact(image_paths, output_path):
    # 1. 加载图片
    imgs = [Image.open(p) for p in image_paths]
    
    # 2. 获取单张图片的宽高 (假设所有图一样大，取第一张即可)
    w, h = imgs[0].size
    
    # 3. 创建画布：宽高各翻一倍
    canvas = Image.new('RGB', (w * 2, h * 2))
    
    # 4. 粘贴到对应位置
    # 第一行
    canvas.paste(imgs[0], (0, 0))   # 左上
    canvas.paste(imgs[1], (w, 0))   # 右上
    
    # 第二行
    canvas.paste(imgs[2], (0, h))   # 左下
    canvas.paste(imgs[3], (w, h))   # 右下
    
    # 5. 保存结果
    canvas.save(output_path, quality=95)
    print(f"拼接成功！输出尺寸为: {w*2}x{h*2}")

# 使用方法：
paths = ['TKE(a).png', 'TKE(b).png', 'TKE(c).png', 'TKE(d).png']

merge_2x2_exact(paths, 'Fig7.jpg')
```
